```python
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
with open("loss_per_iter_loft_vit_r8_loft.txt", "r") as f:
    loss_per_iter = [float(line.strip()) for line in f]
with open("loss_per_iter_lora_vit_r16.txt", "r") as f:
    loss_per_iter_lora = [float(line.strip()) for line in f]    
plt.figure(figsize=(10, 6))
plt.plot(loss_per_iter, label="LoFT ViT r=8")
plt.plot(loss_per_iter_lora, label="LoRA ViT r=8")
plt.xlabel("Iteration")
plt.ylabel("Loss")
plt.title("Loss per Iteration for LoFT ViT r=8 and LoRA ViT r=8")
plt.legend()
plt.grid()
```
